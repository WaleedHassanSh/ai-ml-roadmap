## Task 18 — Manual Bubble Sort

**Output:**

```text
[1, 3, 5, 7, 9]
```

**Lesson:** Compare each value with its right adjacent value and bubble the larger values toward the end.


````python
arr = [9, 3, 7, 1, 5]

n = len(arr)

for i in range(n):
    for j in range(n - i - 1):
        if arr[j] > arr[j + 1]:
            tmp = arr[j]
            arr[j] = arr[j + 1]
            arr[j + 1] = tmp

print(arr)
````

## Task 19 — Manual Selection Sort

**Output:**

```text
[2, 4, 6, 8, 10]
```

**Lesson:** Select the smallest value from the unsorted part and swap it with the current position.


````python
arr = [10, 4, 8, 2, 6]

n = len(arr)

for i in range(n - 1):
    min = arr[i]
    index = i

    for j in range(i + 1, n):
        if min > arr[j]:
            min = arr[j]
            index = j

    tmp = arr[i]
    arr[i] = arr[index]
    arr[index] = tmp

print(arr)
````

## Task 20 — Merge Two Sorted Lists

**Output:**

```text
[1, 2, 3, 4, 7, 8]
```

**Lesson:** Compare values from both lists, select the smaller one, and append it to the merged list.


````python
list1 = [1, 4, 7]
list2 = [2, 3, 8]
list3 = []

n = len(list1)
m = len(list2)

i, j = 0, 0

while i < n and j < m:
    if list1[i] < list2[j]:
        list3.append(list1[i])
        i += 1
    else:
        list3.append(list2[j])
        j += 1

while i < n:
    list3.append(list1[i])
    i += 1

while j < m:
    list3.append(list2[j])
    j += 1

print(list3)
````

## Task 25 — Factorial Calculation

**Output:**

```text
120
```

**Lesson:** Use a recursive call to calculate factorial.


````python
def fact(num):
    if num <= 1:
        return 1

    return num * fact(num - 1)


def main():
    num = 5

    print(fact(num))


main()
````

## Task 26 — Fibonacci Sequence

**Output:**

```text
[0, 1, 1, 2, 3, 5, 8]
```

**Lesson:** Generate the Fibonacci sequence using iteration.


````python
def fact(num):
    a = 0
    b = 1

    factorial = []

    for i in range(num):
        factorial.append(a)

        c = a
        a = b
        b = b + c

    return factorial


def main():
    num = 7

    print(fact(num))


main()
````

## Task 30 — Mini Menu Logic

**Output:**

```text

```

**Lesson:** Create a menu using a loop, choices, conditions, and a list to store data.


````python
stored_data = []

while True:
    try:
        choice = int(
            input(
                """1. Add data
2. View data
3. Exit\n"""
            )
        )

        if choice not in [1, 2, 3]:
            print("\nEnter valid choice please\n")
            continue

        if choice == 1:
            stored_data.append(input("Data entered: "))

        elif choice == 2:
            print(stored_data)

        elif choice == 3:
            break

    except ValueError:
        print("\nEnter valid choice please\n")
````
